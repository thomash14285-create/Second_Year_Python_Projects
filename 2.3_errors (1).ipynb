```python
#Exercise 6

import numpy as np
Lx=uc.ufloat(10.2,0.2)
Ly=uc.ufloat(5.2,0.3)
Lz=uc.ufloat(20,0.1)
T=uc.ufloat(23.1,1)
c=331.3+(T*0.606)

nx1=1
ny1=1
nz1=1

nx2=1
ny2=1
nz2=2

nx3=2
ny3=1
nz3=1

f1=(0.5*c)*(((nx1**2)/(Lx**2)) + ((ny1)**2)/(Ly**2) + ((nz1)**2)/(Lz**2))**0.5
f2=(0.5*c)*(((nx2**2)/(Lx**2)) + ((ny2)**2)/(Ly**2) + ((nz2)**2)/(Lz**2))**0.5
f3=(0.5*c)*(((nx3**2)/(Lx**2)) + ((ny3)**2)/(Ly**2) + ((nz3)**2)/(Lz**2))**0.5

print('c is:', c, 'm/s')
print('f1 is:', f1, 'Hz')
print('f2 is:', f2, 'Hz')
print('f3 is:', f3, 'Hz')
```

```python
#Exercise 7

thetai=uc.ufloat(78,1)
thetairad=thetai*2*np.pi/360
thetat=uc.ufloat(40,1)
thetatrad=thetat*2*np.pi/360


def Rp(thetai, thetat):
    rp = (um.tan(thetai-thetat)/um.tan(thetai+thetat))
    return rp**2

print('R parallel is:', Rp(thetairad, thetatrad))



```

```python
#Exercise 8

mdot=uc.ufloat(0.13,0.01)
d=uc.ufloat(0.011,0.001)
D=uc.ufloat(0.071,0.001)
row=uc.ufloat(0.00101,0.00001)
dP=uc.ufloat(156,7)

#or i.e. -> D = 1e3*uc.ufloat(1.01,0.01)  (whcih converts to SI)


def Cd(mdot,d,D,row,dP):
    return mdot*((1-((d/D)**4)**0.5)/d**2*(row*dP)**0.5)

Cd_value=Cd(mdot,d,D,row,dP)
frac_error= Cd_value.std_dev / Cd_value.nominal_value

print('Fractional error in Cd {:.0%}'.format(frac_error))

#'{:.0%}' outputs the percentage (of frac_error) to 0 significant figures
#frac_error tells Python to put the value of frac_error into the placeholder {:.0%}

```
